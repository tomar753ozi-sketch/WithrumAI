```python
import torch_xla.core.xla_model as xm
import torch_xla.distributed.xla_multiprocessing as xmp

device = xm.xla_device()
print(f'TPU: {device}')
```

```python
import sys
sys.path.insert(0, '/kaggle/input/whitrum-code/WithrumAI')

from whitrum import WhitrumConfig, WhitrumForCausalLM

config = WhitrumConfig(
    vocab_size=60006,
    hidden_size=512,
    intermediate_size=2048,
    num_hidden_layers=12,
    num_attention_heads=8,
    num_key_value_heads=2,
    max_position_embeddings=4096,
)

model = WhitrumForCausalLM(config)
model = model.to(device)
print(f'Params: {sum(p.numel() for p in model.parameters()) / 1e6:.1f}M')
```

```python
from transformers import AutoTokenizer

try:
    tokenizer = AutoTokenizer.from_pretrained('Qwen/Qwen2.5-0.5B', trust_remote_code=True)
except:
    tokenizer = AutoTokenizer.from_pretrained('gpt2')

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print(f'Vocab: {tokenizer.vocab_size}')
```

```python
import json
from datasets import Dataset

with open('/kaggle/input/whitrum-data/training_60k.jsonl') as f:
    data = [json.loads(line) for line in f]

ds = Dataset.from_list(data)
print(f'Dataset: {len(ds)} rows')
```

```python
def tokenize_and_count(texts, tokenizer, max_length=512):
    all_ids = []
    for text in texts:
        ids = tokenizer.encode(text, add_special_tokens=False)
        all_ids.extend(ids)
    return all_ids

all_tokens = tokenize_and_count([d['text'] for d in data], tokenizer)
print(f'Total tokens: {len(all_tokens)}')
```

```python
CHUNK_SIZE = 80_000_000  # 80M tokens per chunk
NUM_CHUNKS = 5
TOTAL_TOKENS = CHUNK_SIZE * NUM_CHUNKS

chunks = []
for i in range(NUM_CHUNKS):
    start = i * CHUNK_SIZE
    end = min(start + CHUNK_SIZE, len(all_tokens))
    chunk = all_tokens[start:end]
    chunks.append(chunk)
    print(f'Chunk {i+1}: {len(chunk)} tokens')

print(f'\nTotal: {sum(len(c) for c in chunks)} tokens in {len(chunks)} chunks')
```

```python
import torch
from torch.utils.data import DataLoader, Dataset as TorchDataset
import os

class ChunkDataset(TorchDataset):
    def __init__(self, token_ids, seq_len=512):
        self.seq_len = seq_len
        self.data = []
        for i in range(0, len(token_ids) - seq_len, seq_len):
            self.data.append(token_ids[i:i+seq_len])
    def __len__(self):
        return len(self.data)
    def __getitem__(self, idx):
        ids = self.data[idx]
        return {
            'input_ids': torch.tensor(ids, dtype=torch.long),
            'attention_mask': torch.ones(len(ids), dtype=torch.long),
            'labels': torch.tensor(ids, dtype=torch.long),
        }
```

```python
from torch.optim import AdamW
import time

optimizer = AdamW(model.parameters(), lr=2e-4)
num_epochs = 2
save_dir = '/kaggle/working/checkpoints'
os.makedirs(save_dir, exist_ok=True)

for chunk_idx, chunk_tokens in enumerate(chunks):
    print(f'\n=== Chunk {chunk_idx+1}/{NUM_CHUNKS} ({len(chunk_tokens)} tokens) ===')
    
    chunk_dataset = ChunkDataset(chunk_tokens, seq_len=512)
    chunk_loader = DataLoader(chunk_dataset, batch_size=8, shuffle=True)
    
    print(f'Steps per chunk: {len(chunk_loader)}')
    
    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        start_time = time.time()
        
        for i, batch in enumerate(chunk_loader):
            batch = {k: v.to(device) for k, v in batch.items()}
            
            outputs = model(**batch)
            loss = outputs.loss
            
            loss.backward()
            xm.optimizer_step(optimizer)
            optimizer.zero_grad()
            
            total_loss += loss.item()
            
            if i % 50 == 0:
                elapsed = time.time() - start_time
                print(f'  Step {i}/{len(chunk_loader)} | Loss: {loss.item():.4f} | Time: {elapsed:.0f}s')
        
        avg_loss = total_loss / len(chunk_loader)
        print(f'  Epoch {epoch+1} done | Avg Loss: {avg_loss:.4f}')
    
    # Save checkpoint after each chunk
    ckpt_path = f'{save_dir}/chunk_{chunk_idx+1}.pt'
    torch.save({
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'chunk': chunk_idx + 1,
        'loss': avg_loss,
    }, ckpt_path)
    print(f'  Checkpoint saved: {ckpt_path}')
```

```python
final_path = '/kaggle/working/whitrum-350m-final.pt'
torch.save({
    'model_state_dict': model.state_dict(),
    'config': config.to_dict(),
    'total_chunks': NUM_CHUNKS,
    'total_tokens': TOTAL_TOKENS,
}, final_path)
print(f'Final model saved: {final_path}')
```
